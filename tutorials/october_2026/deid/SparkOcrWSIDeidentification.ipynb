## Whole Slide Imaging De-identification

In this notebook, we will explore how to handle the de-identification of Whole Slide Imaging (WSI) files, specifically focusing on SVS files.</br>
We will cover:
* Removal of auxiliary images, like macro and label images.
* Removal of Metadata.
* Main image cleaning: in this situation we don't want to remove the entire image like the case of for example macro or label images, but we want to remove only specific regions which contain PHI. We want to do this keeping the quality of the original image intact. This process involves the following steps which we will discuss in next sections,
  * Extract from the SVS an image of the whole slide, at low resolution, and its thumbnail
  * De-identify those images
  * Update the original source SVS file to mask the regions containing PHI, on every level.

Let's jump right in!

### Introduction to the SVS file format

The SVS file format (commonly used for whole slide images (WSI) in digital pathology) is based on the TIFF format and supports multi-resolution (pyramidal) image storage. It allows storing extremely large images efficiently, along with metadata and multiple resolutions for fast zooming and viewing.
SVS typically stores images in a pyramid of resolutions,
* level 0 -> full resolution.
* level 1 -> downsampled version, like 1/4 of the resolution at level 0.
* ...
* level N -> more downsampling.

Something like this(image taken from nema.org),</br>
![sample wsi](https://dicom.nema.org/dicom/dicomwsi/sup145_fromword_files/image008.gif)

At each level, images are not stored as a single monolithic unit, but as a set of _tiles_. This enables for a more efficient handling of the image by viewers.</br>
Text found in the scanned area, like notes written with a pen on the glass or the edge of the slide label, is millimeters high: it is looked for on a low resolution view of the whole slide, and the regions found are then blacked out tile by tile on every level.

### Removal of auxiliary images, like macro and label images, along with metadata.
First things first, before diving into the full resolution images, let's see how to remove the auxiliary images and the metadata. For this, the `svs.remove_phi()` comes handy. Let's take a look at the documentation,

```python
from sparkocr.utils.svs.phi_cleaning import remove_phi
help(remove_phi)
```

So we see from the documentation that we can remove label, and macro images along with a predefined set of metadata tags. We can also add our own custom list of tags.</br>
We have everything we need, so let's do it!,

```python
# make sure input files are in this folder
input_path = './data/wsi/wsi-sample.svs'
output_path = "deid_svs"

remove_phi(input_path, output_path, verbose=True)
```

```python
input_path = './data/wsi/wsi-sample.svs'
output_path = "svs_output"

remove_phi(input_path, output_path, verbose=True, append_tags=['ImageDescription.ScanScope ID', 'ImageDescription.Time Zone', 'ImageDescription.ScannerType'])
```

### Cleaning of the main image
Here we will de-identify the main image. We start by extracting from the SVS the images to look for PHI in: the whole slide at low resolution, so that no text is cut between tiles, and its thumbnail.

```python
from sparkocr.utils.svs.deidentify import svs_to_text_images

OUTPUT_FOLDER = "text_output/"
images = svs_to_text_images("deid_svs/wsi-sample.svs", OUTPUT_FOLDER)
images
```

#### Inspect the images

```python
from sparkocr.transformers import *
from sparkocr.enums import *
from sparkocr.utils import display_images

bin_to_image = BinaryToImage() \
    .setInputCol("content") \
    .setOutputCol("image_raw") \
    .setImageType(ImageType.TYPE_BYTE_GRAY)

image_df = bin_to_image.transform(spark.read.format("binaryFile").load(images))
display_images(image_df, "image_raw")
```

#### PHI removal per-se
We use the image de-identification pipeline: it detects the text, reads it (OCR) and finds the PHI in it (NER).</br>
The OCR returns the text of the slide line by line, so we make each line a sentence: this way an entity never runs from one line into the next.

```python
from sparkocr.pretrained import PretrainedPipeline

deid_pipeline = PretrainedPipeline("image_deid_multi_model_context_pipeline_cpu", "en", "clinical/ocr").model
[stage for stage in deid_pipeline.stages if type(stage).__name__ == "SentenceDetectorDLModel"][0].setCustomBounds(["\n"])
```

### Run pipeline

```python
result_deid = deid_pipeline.transform(image_df)
deid_info = result_deid.select("path", "coordinates")
```

```python
deid_info.show(5, False)
```

#### [optional] Cache regions using parquet
We can cache the regions in disk, to avoid potential re-computation and to be able to recover in case of failure.

```python
deid_info.repartition(10).write.format("parquet").mode("overwrite").save("./cached_regions.parquet")
```

```python
deid_info = spark.read.parquet("./cached_regions.parquet")
```

#### Display the detected PHI
Let's take a look at the result of this process

```python
display_images(result_deid, "image_with_regions")
```

#### Masking of the regions in the SVS
This is our final step!. We will mask the regions we obtained during previous steps on every level of the SVS and on its thumbnail.</br>
With `create_new_svs_file=False` the SVS is modified in place, so first let's create a copy of our files.

```python
!mkdir -p deid_svs_pipeline && cp deid_svs/wsi-sample.svs deid_svs_pipeline/

from sparkocr.utils.svs.deidentify import redact_phi_in_text_images

# the SVS is modified in place
redact_phi_in_text_images("deid_svs_pipeline/wsi-sample.svs", deid_info, OUTPUT_FOLDER, create_new_svs_file=False);
```

### Blanket de-identification
To black out all the text on the slide, PHI or not, we use a pipeline that only detects the text, instead of the de-identification pipeline. The rest of the steps are the same.

```python
from pyspark.ml import PipelineModel

text_detector = ImageTextDetector.pretrained("image_text_detector_mem_opt", "en", "clinical/ocr") \
    .setInputCol("image_raw") \
    .setOutputCol("coordinates") \
    .setScoreThreshold(0.7) \
    .setLinkThreshold(0.5) \
    .setWithRefiner(True) \
    .setTextThreshold(0.4) \
    .setSizeThreshold(-1) \
    .setWidth(0) \
    .setHeight(0)

text_detection = PipelineModel(stages=[text_detector])
text_info = text_detection.transform(image_df).select("path", "coordinates")
```

```python
# create_new_svs_file=True: a new SVS file is created at output_svs_path, which takes a bit longer
redact_phi_in_text_images("deid_svs/wsi-sample.svs", text_info, OUTPUT_FOLDER, output_svs_path="deid_svs_blanket/wsi-sample.svs", create_new_svs_file=True);
```
